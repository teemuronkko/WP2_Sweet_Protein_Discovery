```python
import os 

# Get directory of the script
script_dir = os.path.dirname(os.getcwd())
repo_name = "WP2_SweetProteinDiscovery"
repo_dir = script_dir.split(repo_name)[0] + repo_name

onekp_path = f"{repo_dir}/5_openfold3_models/onekp_hits/predictions"
uniprot_path = f"{repo_dir}/5_openfold3_models/uniprot_missing_structures/predictions"

```

```python
# List files in onekp_path
temp_dir = "./temp/"
fail_c = 0 
onekp_files = os.listdir(uniprot_path)
print("Files in UniProt path:")
for file in onekp_files:
    # List files in subdir
    subdir_path = os.path.join(uniprot_path, file)
    if os.path.isdir(subdir_path):

        # Check if "summary.txt" is in subdir_path
        summary_path = os.path.join(subdir_path, "summary.txt")
        if not os.path.exists(summary_path):
            print(f"summary.txt not found in {file}.")
            #os.makedirs(os.path.join(temp_dir, file), exist_ok=True)
            #os.rename(subdir_path, os.path.join(temp_dir, file))
            fail_c += 1
            continue

        model_dir = f"{subdir_path}/{file}.temp/seed_42"


        if not os.path.exists(model_dir):
            print(f"  Files in {file}:")
            print(f"    Moved {file} to temp directory.")
            os.makedirs(os.path.join(temp_dir, file), exist_ok=True)
            os.rename(subdir_path, os.path.join(temp_dir, file))
            fail_c += 1
            continue
        else:
            subdir_files = os.listdir(model_dir)
            # Remove hidden files
            subdir_files = [f for f in subdir_files if not f.startswith('.')]

            # Check if no pdb files are present
            pdb_files = [f for f in subdir_files if f.endswith('.pdb')]
            if len(pdb_files) == 0:
                print(f"    Moved {file} to temp directory.")
                print(f"    No PDB files found.")
                os.makedirs(os.path.join(temp_dir, file), exist_ok=True)
                os.rename(subdir_path, os.path.join(temp_dir, file))
                fail_c += 1
                continue

            if len(subdir_files) != 21 and len(subdir_files) != 16:
                print(f"    Moved {file} to temp directory.")
                print(f"    Expected 21 files, found {len(subdir_files)} files.")
                os.makedirs(os.path.join(temp_dir, file), exist_ok=True)
                os.rename(subdir_path, os.path.join(temp_dir, file))
                fail_c += 1
                continue

            # If "timing.json" is not in files, move subdir to temp_dir
            if "timing.json" not in subdir_files:
                os.makedirs(os.path.join(temp_dir, file), exist_ok=True)
                os.rename(subdir_path, os.path.join(temp_dir, file))
                fail_c += 1
                print(f"    Moved {file} to temp directory.")
```
