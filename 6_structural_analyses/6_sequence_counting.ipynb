```python
import pandas as pd
import os

repo_dir = "WP2_Sweet_Protein_Discovery"
repo_path = os.path.join(os.getcwd().split(repo_dir)[0], repo_dir)
print(f"Repository path: {repo_path}")

all_tm_scores = pd.read_csv(f"{repo_path}/6_structural_analyses/1_tm_scores/all_tm_scores.tsv", sep="\t")

# Drop pident 100.0 and fident 1.00 rows
all_tm_scores = all_tm_scores[(all_tm_scores["pident"] != 100.0) & (all_tm_scores["fident"] != 1.00)] 
print(f"Number of rows after dropping pident 100.0 and fident 1.00: {len(all_tm_scores)}")

# Keep only one row per target
all_tm_scores = all_tm_scores.sort_values(by="pident", ascending=False).drop_duplicates(subset=["target"], keep="first")
print(f"Number of rows after keeping only one row per target: {len(all_tm_scores)}")

len(all_tm_scores)
```

```python
# Count number of fastas in directory
fasta_dir = f"{repo_path}/6_structural_analyses/3_representative_structures/fastas/peptides"
# Count the number of fasta files in the directory
num_fastas = len([f for f in os.listdir(fasta_dir) if f.endswith('.fasta')])
print(f"Number of fasta files in {fasta_dir}: {num_fastas}")
```

```python
of3_json = f"{repo_path}/6_structural_analyses/3_representative_structures/of3_jsons/complexes" 
# Count the number of json files in the directory
num_of3_jsons = len([f for f in os.listdir(of3_json) if f.endswith('.json')])
print(f"Number of json files in {of3_json}: {num_of3_jsons}")

monomer_jsons = f"{repo_path}/6_structural_analyses/3_representative_structures/of3_jsons/monomers"

num_monomer_jsons = len([f for f in os.listdir(monomer_jsons) if f.endswith('.json')])
print(f"Number of json files in {monomer_jsons}: {num_monomer_jsons}")

# Find out which complexes are not in monomer jsons
not_in_monomer_jsons = []
for complex_json in os.listdir(of3_json):
    if complex_json.endswith('.json'):
        monomer_json = complex_json.replace('complex', 'monomer')
        if not os.path.exists(os.path.join(monomer_jsons, monomer_json)):
            not_in_monomer_jsons.append(complex_json)

not_in_monomer_jsons
```
